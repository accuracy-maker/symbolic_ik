## Power of Symbolic Computation
Sympy can simplify expressions, compute derivatives, integrals and limits, solve equations, work with matrices etc.

```python
from sympy import *
x, t, z, nu = symbols('x t z nu')
# this makes all further examples pretty print
init_printing(use_unicode=True)
```

```python
diff(sin(x)*exp(x), x)
```

```python
integrate(exp(x)*sin(x) + exp(x)*cos(x), x)
```

```python
integrate(sin(x**2), (x, -oo, oo))
```

```python
limit(sin(x)/x, x, 0)
```

```python
solve(x**2 - 2, x)
```

```python
y = Function('y')
dsolve(Eq(y(t).diff(t, t) - y(t), exp(t)), y(t))
```

```python
# find the eigenvalues of [[1,2],[2,2]]
Matrix([[1, 2], [2, 2]]).eigenvals()
```

```python
besselj(nu, z).rewrite(jn)
```

```python
latex(Integral(cos(x)**2, (x, 0, pi)))
```

## Polynomial System

```python
x, y, z = symbols('x,y,z')
```

```python
# coefficients are integers
e = (x + y)*(y - 2*z)
e.as_poly()
```

```python
# coefficients are rational number
e = (3*x/2 + y)*(z - 1)
e.as_poly()
```

```python
e = (x + 2*pi)*y
e.as_poly()
```

```python
e.as_poly(x, y)
```

```python
# solve polynomial equation
solve(x**3 + 2*x + 3, x)
```

```python
p = Symbol('p')
q = Symbol('q')
solve(x**2 + p*x + q, x)
```

```python
solve_poly_system([y - x, x - 5], x, y)
```

```python
solve_poly_system([y**2 - x**3 + 1, y*x], x, y)
```
